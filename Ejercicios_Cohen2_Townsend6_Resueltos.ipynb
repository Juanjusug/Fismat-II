## Bloque A: Cohen, Capitulo 2

### Ejercicio A1 - Normalizacion y probabilidades

**Enunciado**

Sea

$$
|\psi\rangle = \frac{1}{\sqrt{14}}\left(|1\rangle + 2|2\rangle + 3|3\rangle\right),
$$

en la base ortonormal $\{|1\rangle, |2\rangle, |3\rangle\}$.

1. Verificar normalizacion.
2. Hallar $P(1)$, $P(2)$ y $P(3)$.

**Solucion**

$$
\langle \psi|\psi \rangle = \frac{1}{14}(1^2 + 2^2 + 3^2) = \frac{14}{14} = 1.
$$

El estado esta normalizado. Las probabilidades son:

$$
P(1)=\left|\frac{1}{\sqrt{14}}\right|^2=\frac{1}{14},
\quad
P(2)=\left|\frac{2}{\sqrt{14}}\right|^2=\frac{4}{14}=\frac{2}{7},
\quad
P(3)=\left|\frac{3}{\sqrt{14}}\right|^2=\frac{9}{14}.
$$

Comprobacion: $P(1)+P(2)+P(3)=1$.

### Ejercicio A2 - Valor esperado y dispersion

**Enunciado**

Con el estado del ejercicio A1, considere

$$
\hat A = a|1\rangle\langle1| + 2a|2\rangle\langle2| + 4a|3\rangle\langle3|.
$$

Calcular $\langle A \rangle$, $\langle A^2 \rangle$ y $\Delta A$.

**Solucion**

$$
\langle A \rangle = aP(1)+2aP(2)+4aP(3)
= a\left(\frac{1}{14}+\frac{8}{14}+\frac{36}{14}\right)
= \frac{45}{14}a.
$$

Para $\hat A^2$ (autovalores $a^2,4a^2,16a^2$):

$$
\langle A^2 \rangle = a^2P(1)+4a^2P(2)+16a^2P(3)
= a^2\left(\frac{1}{14}+\frac{16}{14}+\frac{144}{14}\right)
= \frac{161}{14}a^2.
$$

$$
\Delta A = \sqrt{\langle A^2 \rangle - \langle A \rangle^2}
= a\sqrt{\frac{229}{196}}
= \frac{\sqrt{229}}{14}a.
$$

### Ejercicio A3 - Conmutador y compatibilidad

**Enunciado**

En una base de dos estados:

$$
A = \begin{pmatrix}1 & 0\\0 & -1\end{pmatrix},
\quad
B = \begin{pmatrix}0 & 1\\1 & 0\end{pmatrix}.
$$

1. Hallar $[A,B]=AB-BA$.
2. Decidir si son compatibles.

```python
import sympy as sp
A = sp.Matrix([[1, 0], [0, -1]])
B = sp.Matrix([[0, 1], [1, 0]])
comm = A*B - B*A
A, B, comm
```

**Solucion**

$$
[A,B] = \begin{pmatrix}0 & 2\\-2 & 0\end{pmatrix} \neq 0.
$$

No conmutan, por lo tanto no son observables compatibles.

### Ejercicio A4 - Proyeccion y colapso

**Enunciado**

Sea

$$
|\psi\rangle = \frac{1}{\sqrt{5}}(2|u\rangle + i|v\rangle),
$$

con $\langle u|u\rangle=\langle v|v\rangle=1$ y $\langle u|v\rangle=0$.

1. Hallar la probabilidad de medir $P_u=|u\rangle\langle u|$.
2. Hallar el estado posterior para ese resultado.

**Solucion**

$$
P(u)=\langle\psi|P_u|\psi\rangle
=|\langle u|\psi\rangle|^2
=\left|\frac{2}{\sqrt{5}}\right|^2
=\frac{4}{5}.
$$

Estado posterior normalizado:

$$
|\psi'\rangle = \frac{P_u|\psi\rangle}{\sqrt{P(u)}} = |u\rangle.
$$

## Bloque B: Townsend, Capitulo 6

### Ejercicio B1 - Algebra de momento angular

**Enunciado**

Usando

$$
[J_i,J_j] = i\hbar\,\varepsilon_{ijk}J_k,
$$

demostrar que

$$
[J^2,J_z]=0,
\quad
J^2=J_x^2+J_y^2+J_z^2.
$$

**Solucion**

$$
[J^2,J_z]=[J_x^2,J_z]+[J_y^2,J_z]+[J_z^2,J_z].
$$

El tercer termino es cero. Con $[AB,C]=A[B,C]+[A,C]B$:

$$
[J_x^2,J_z]=J_x[J_x,J_z]+[J_x,J_z]J_x=-i\hbar(J_xJ_y+J_yJ_x),
$$

$$
[J_y^2,J_z]=J_y[J_y,J_z]+[J_y,J_z]J_y=+i\hbar(J_yJ_x+J_xJ_y).
$$

La suma se anula, por tanto $[J^2,J_z]=0$.

### Ejercicio B2 - Operadores de escalera

**Enunciado**

Sea $|j,m\rangle$ autoestado de $J^2$ y $J_z$. Hallar la accion de $J_+$ y $J_-$.

**Solucion**

Definiciones:

$$
J_+ = J_x + iJ_y,
\quad
J_- = J_x - iJ_y.
$$

Acciones:

$$
J_+|j,m\rangle = \hbar\sqrt{j(j+1)-m(m+1)}\,|j,m+1\rangle,
$$

$$
J_-|j,m\rangle = \hbar\sqrt{j(j+1)-m(m-1)}\,|j,m-1\rangle.
$$

Estados extremos:

$$
J_+|j,j\rangle=0,
\quad
J_-|j,-j\rangle=0.
$$

Por ello $m=-j,-j+1,\dots,j-1,j$.

### Ejercicio B3 - Medicion de spin en direccion arbitraria

**Enunciado**

El sistema inicia en $|+\rangle_z$ y se mide spin en una direccion $\hat n$ que forma angulo $\theta$ con el eje $z$.

Hallar $P(+_n)$ y $P(-_n)$.

**Solucion**

Para spin 1/2:

$$
|+\rangle_n = \cos\frac{\theta}{2}|+\rangle_z + e^{i\phi}\sin\frac{\theta}{2}|-\rangle_z.
$$

Luego:

$$
P(+_n)=|\langle +_n|+_z\rangle|^2 = \cos^2\frac{\theta}{2},
\quad
P(-_n)=\sin^2\frac{\theta}{2}.
$$

Caso especial: si $\theta=\pi/2$, entonces $P(+_n)=P(-_n)=1/2$.

```python
import sympy as sp
theta = sp.symbols('theta', real=True)
P_plus = sp.cos(theta/2)**2
P_minus = sp.sin(theta/2)**2
sp.simplify(P_plus + P_minus)
```

### Ejercicio B4 - Matrices de Pauli

**Enunciado**

Con

$$
\sigma_x = \begin{pmatrix}0 & 1\\1 & 0\end{pmatrix},
\quad
\sigma_y = \begin{pmatrix}0 & -i\\i & 0\end{pmatrix},
\quad
\sigma_z = \begin{pmatrix}1 & 0\\0 & -1\end{pmatrix},
$$

verificar:

$$
[\sigma_i,\sigma_j]=2i\,\varepsilon_{ijk}\sigma_k,
\quad
\{\sigma_i,\sigma_j\}=2\delta_{ij}I.
$$

```python
I = sp.I
sx = sp.Matrix([[0, 1], [1, 0]])
sy = sp.Matrix([[0, -I], [I, 0]])
sz = sp.Matrix([[1, 0], [0, -1]])
comm_xy = sx*sy - sy*sx
anticomm_xy = sx*sy + sy*sx
comm_xy, anticomm_xy
```

**Solucion**

Identidades utiles:

$$
[\sigma_x,\sigma_y]=2i\sigma_z,
\quad
[\sigma_y,\sigma_z]=2i\sigma_x,
\quad
[\sigma_z,\sigma_x]=2i\sigma_y.
$$

$$
\{\sigma_x,\sigma_x\}=2I,
\quad
\{\sigma_x,\sigma_y\}=0
$$

y analogas por simetria ciclica. Con $S_i=(\hbar/2)\sigma_i$ se recupera el algebra de spin.

### Ejercicio B5 - Numero de estados para un $j$ dado

**Enunciado**

Para un valor fijo de $j$, determinar cuantos valores de $m$ existen y la dimension del multiplete.

**Solucion**

Los valores permitidos son:

$$
m=-j,-j+1,\dots,j-1,j.
$$

Por tanto, el numero total de estados es:

$$
N=2j+1.
$$

Ejemplos:
- $j=0 \Rightarrow N=1$
- $j=1/2 \Rightarrow N=2$
- $j=1 \Rightarrow N=3$
- $j=3/2 \Rightarrow N=4$

### Ejercicio A5 - Valor esperado en estado general de dos niveles
**Enunciado**

Sea

$$
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle,
\quad |\alpha|^2+|\beta|^2=1,
$$

y

$$
\hat A = a|0\rangle\langle0| + b|1\rangle\langle1|.
$$

Hallar $\langle A\rangle$ y $\Delta A$.

**Solucion**

$$
\langle A\rangle = a|\alpha|^2 + b|\beta|^2,
$$

$$
\langle A^2\rangle = a^2|\alpha|^2 + b^2|\beta|^2,
$$

$$
\Delta A = \sqrt{\langle A^2\rangle-\langle A\rangle^2}=|a-b|\,|\alpha|\,|\beta|.
$$

### Ejercicio A6 - Fase global vs fase relativa
**Enunciado**

Comparar:

$$
|\psi_1\rangle = \frac{|0\rangle+|1\rangle}{\sqrt2},
\quad
|\psi_2\rangle = e^{i\chi}|\psi_1\rangle,
\quad
|\psi_3\rangle = \frac{|0\rangle+e^{i\phi}|1\rangle}{\sqrt2}.
$$

**Solucion**

$|\psi_1\rangle$ y $|\psi_2\rangle$ son fisicamente equivalentes (fase global). En cambio $|\psi_3\rangle$ cambia interferencias para $\phi\neq0$ (fase relativa observable).

### Ejercicio A7 - Operador proyector idempotente
**Enunciado**

Para $P_u=|u\rangle\langle u|$ con $\langle u|u\rangle=1$, demostrar $P_u^2=P_u$ y $P_u^\dagger=P_u$.

**Solucion**

$$
P_u^2 = |u\rangle\langle u|u\rangle\langle u| = |u\rangle\langle u| = P_u,
$$

$$
P_u^\dagger=(|u\rangle\langle u|)^\dagger = |u\rangle\langle u| = P_u.
$$

### Ejercicio A8 - Conmutador canonico y Heisenberg
**Enunciado**

Si $[\hat x,\hat p]=i\hbar$, deducir cota de incertidumbre.

**Solucion**

$$

$$

### Ejercicio A9 - Base completa y reconstruccion
**Enunciado**

Sea base ortonormal $\{|n\rangle\}$ con

**Solucion**

### Ejercicio A12 - Condicion de compatibilidad
**Enunciado**

Si $[A,B]=0$ y espectro no degenerado de $A$, mostrar que autoestados de $A$ son autoestados de $B$.

**Solucion**

Para $A|a_n\rangle=a_n|a_n\rangle$:

$$
A(B|a_n\rangle)=BA|a_n\rangle=a_n(B|a_n\rangle).
$$

Como el subespacio de $a_n$ es 1D (no degenerado), $B|a_n\rangle\propto|a_n\rangle$.

### Ejercicio A13 - Descomposicion espectral
**Enunciado**

Para observable discreto no degenerado $A$, escribir forma espectral.

**Solucion**

$$
A=\sum_n a_n|a_n\rangle\langle a_n|,
\quad
f(A)=\sum_n f(a_n)|a_n\rangle\langle a_n|.
$$

### Ejercicio A14 - Matriz densidad de estado puro
**Enunciado**

Sea $\rho=|\psi\rangle\langle\psi|$. Mostrar que $\rho^2=\rho$ y $\mathrm{Tr}(\rho)=1$.

**Solucion**

$$
\rho^2=|\psi\rangle\langle\psi|\psi\rangle\langle\psi|=|\psi\rangle\langle\psi|=\rho,
\quad
\mathrm{Tr}(\rho)=\langle\psi|\psi\rangle=1.
$$

### Ejercicio A15 - Probabilidad en subespacio degenerado
**Enunciado**

Si $P_\lambda$ proyecta al subespacio de autovalor degenerado $\lambda$, hallar probabilidad y estado postmedicion.

**Solucion**

$$
p(\lambda)=\langle\psi|P_\lambda|\psi\rangle,
\quad
|\psi'\rangle=\frac{P_\lambda|\psi\rangle}{\sqrt{p(\lambda)}}.
$$

### Ejercicio B25 - Cierre de serie de Townsend
**Enunciado**: Verificar normalizacion de probabilidades para cualquier direccion de medicion.

**Solucion**:

$$

$$

Con esto, la serie de Townsend queda extendida hasta 25 ejercicios (B1-B25).

```python
# Verificacion simbolica rapida de identidades usadas en la ampliacion
import sympy as sp

hbar = sp.symbols('hbar', positive=True, real=True)
theta = sp.symbols('theta', real=True)

# Probabilidades angulares de spin 1/2
P_plus = sp.cos(theta/2)**2
P_minus = sp.sin(theta/2)**2
print('P_plus + P_minus =', sp.simplify(P_plus + P_minus))

# Matrices de Pauli
I = sp.I
sx = sp.Matrix([[0, 1], [1, 0]])
sy = sp.Matrix([[0, -I], [I, 0]])
sz = sp.Matrix([[1, 0], [0, -1]])
Id = sp.eye(2)

# Conmutadores y anticonmutadores
comm_xy = sx*sy - sy*sx
anticomm_xy = sx*sy + sy*sx
print('[sx, sy] =')
sp.pprint(comm_xy)
print('{sx, sy} =')
sp.pprint(anticomm_xy)

# Verificacion de proyector P_{+z} = (I + sz)/2
P_plus_z = (Id + sz)/2
print('P_plus_z^2 - P_plus_z =')
sp.pprint(sp.simplify(P_plus_z*P_plus_z - P_plus_z))
```
