## 1. Motivación e Introducción {#motivacion}

### ¿Por qué necesitamos mecánica cuántica?

La mecánica clásica describe exitosamente el movimiento de cuerpos macroscópicos. Sin embargo, a escalas atómicas y subatómicas:

- **Radiación de cuerpo negro**: Los átomos emiten energía de forma discreta, no continua
- **Efecto fotoeléctrico**: La luz se comporta como partículas (fotones) con energía $E = h\nu$
- **Espectros atómicos discretos**: Los átomos producen líneas espectrales específicas
- **Estabilidad atómica**: Los electrones no colapsan al núcleo como predice el electromagnetismo clásico
- **Comportamiento onda-partícula**: La materia y la radiación tienen propiedades duales

### Conceptos Preliminares

**Relación de de Broglie** (1924):  
Toda partícula con momento $p$ tiene asociada una longitud de onda:

$$\lambda = \frac{h}{p} = \frac{\hbar}{p}$$

donde $\hbar = \frac{h}{2\pi} = 1.054571817 \times 10^{-34}$ J·s es la constante de Planck reducida.

**Cuantización de energía** (Planck, Bohr):  
La energía no es continua sino discreta:

$$E = n\hbar\omega = nh\nu \quad (n = 1, 2, 3, ...)$$

**Principio de incertidumbre de Heisenberg** (1927):  
No es posible conocer simultáneamente con precisión arbitraria la posición y el momento:

$$\Delta x \cdot \Delta p \geq \frac{\hbar}{2}$$

Análogamente para energía y tiempo:

$$\Delta E \cdot \Delta t \geq \frac{\hbar}{2}$$

## 2. Postulados de la Mecánica Cuántica {#postulados}

La mecánica cuántica se fundamenta en un conjunto de postulados que definen su estructura matemática y física.

### Postulado 1: Espacio de Hilbert

**Enunciado**: El estado de un sistema cuántico está completamente descrito por un vector de estado $|\psi\rangle$ perteneciente a un espacio de Hilbert $\mathcal{H}$.

Propiedades:
- Es un espacio vectorial complejo
- Tiene producto interno definido: $\langle\phi|\psi\rangle$
- Es completo (toda sucesión de Cauchy converge)
- Para cada vector $|\psi\rangle$, existe un vector dual $\langle\psi|$ en el espacio dual

### Postulado 2: Observables como Operadores

**Enunciado**: Toda cantidad medible (observable) en mecánica cuántica está representada por un operador lineal hermitiano $\hat{A}$ en el espacio de Hilbert.

Propiedades de operadores hermitianos:
$$\hat{A} = \hat{A}^\dagger$$

donde $\hat{A}^\dagger = (\hat{A}^T)^*$ es el adjunto hermitiano.

**Observables comunes**:
- Posición: $\hat{x}$
- Momento: $\hat{p} = -i\hbar\frac{d}{dx}$
- Energía: $\hat{H}$ (Hamiltoniano)
- Momento angular: $\hat{L} = \hat{r} \times \hat{p}$

### Postulado 3: Valores Propios y Medición

**Enunciado**: Los únicos resultados posibles de una medición del observable $\hat{A}$ son sus valores propios $a_n$, que satisfacen:

$$\hat{A}|a_n\rangle = a_n|a_n\rangle$$

donde $|a_n\rangle$ son los vectores propios (estados propios).

Para operadores hermitianos:
- Los valores propios son **reales**
- Los vectores propios son **ortogonales** (si pertenecen a valores propios diferentes)
- Forman una **base completa** del espacio de Hilbert

### Postulado 4: Probabilidad e Interpretación de Born

**Enunciado**: Si un sistema está en estado $|\psi\rangle$ y se mide el observable $\hat{A}$, la probabilidad de obtener el valor propio $a_n$ es:

$$P(a_n) = |\langle a_n|\psi\rangle|^2$$

El estado $|\psi\rangle$ debe estar normalizado:
$$\langle\psi|\psi\rangle = 1$$

Para un estado general en la base de valores propios de $\hat{A}$:
$$|\psi\rangle = \sum_n c_n|a_n\rangle$$

la probabilidad es:
$$P(a_n) = |c_n|^2$$

con la condición de normalización:
$$\sum_n |c_n|^2 = 1$$

### Postulado 5: Colapso de la Función de Onda

**Enunciado**: Después de medir el observable $\hat{A}$ y obtener el resultado $a_n$, el sistema colapsa al estado propio correspondiente:

$$|\psi\rangle \rightarrow |a_n\rangle$$

Este es un cambio discontinuo que tiene lugar en el momento de la medición.

### Postulado 6: Evolución Temporal

**Enunciado**: La evolución temporal de un sistema cuántico aislado está gobernada por la ecuación de Schrödinger dependiente del tiempo:

$$i\hbar\frac{\partial|\psi\rangle}{\partial t} = \hat{H}|\psi\rangle$$

donde $\hat{H}$ es el operador Hamiltoniano (energía total del sistema).

## 3. Espacios de Hilbert y Notación de Dirac {#hilbert}

### Notación de Dirac (Bra-Ket)

La notación de Dirac simplifica enormemente la manipulación de estados cuánticos:

- **Ket** $|\psi\rangle$: Vector de estado (columna)
- **Bra** $\langle\psi|$: Vector dual (fila, adjunto hermitiano)
- **Producto interno**: $\langle\phi|\psi\rangle$ (número complejo)
- **Producto externo**: $|\psi\rangle\langle\phi|$ (operador)

### Propiedades del Producto Interno

Para vectores $|\psi\rangle, |\phi\rangle \in \mathcal{H}$ y escalares $c, d \in \mathbb{C}$:

1. **Linealidad en el segundo argumento**:
   $$\langle\psi|(c|\phi_1\rangle + d|\phi_2\rangle) = c\langle\psi|\phi_1\rangle + d\langle\psi|\phi_2\rangle$$

2. **Antilinealidad en el primer argumento**:
   $$(c|\psi_1\rangle + d|\psi_2\rangle)^\dagger = c^*\langle\psi_1| + d^*\langle\psi_2|$$

3. **Simetría hermitiana**:
   $$\langle\psi|\phi\rangle = \langle\phi|\psi\rangle^*$$

4. **Positividad**:
   $$\langle\psi|\psi\rangle \geq 0, \quad \langle\psi|\psi\rangle = 0 \Rightarrow |\psi\rangle = 0$$

### Norma y Normalización

La **norma** de un estado es:
$$\|\psi\| = \sqrt{\langle\psi|\psi\rangle}$$

Un estado se dice **normalizado** si $\langle\psi|\psi\rangle = 1$.

Un estado normalizado arbitrario puede escribirse como:
$$|\psi\rangle = \frac{|\psi_0\rangle}{\|\psi_0\|}$$

### Bases Ortonormales

Un conjunto de vectores $\{|e_i\rangle\}$ es una **base ortonormal** si:

1. **Ortonormalidad**:
   $$\langle e_i|e_j\rangle = \delta_{ij} = \begin{cases} 1 & i = j \\ 0 & i \neq j \end{cases}$$

2. **Completitud** (relación de clausura):
   $$\sum_i |e_i\rangle\langle e_i| = \mathbb{I}$$

### Expansión en Base Ortonormal

Todo vector $|\psi\rangle$ puede expandirse como:
$$|\psi\rangle = \sum_i c_i|e_i\rangle$$

donde los coeficientes son:
$$c_i = \langle e_i|\psi\rangle$$

y la norma es:
$$\langle\psi|\psi\rangle = \sum_i |c_i|^2$$

### Operadores en Representación Matricial

En una base ortonormal $\{|e_i\rangle\}$, un operador $\hat{A}$ tiene elementos de matriz:
$$A_{ij} = \langle e_i|\hat{A}|e_j\rangle$$

La acción del operador en un vector es:
$$\hat{A}|\psi\rangle = \sum_i \left(\sum_j A_{ij}c_j\right)|e_i\rangle$$

En forma matricial:
$$\mathbf{A} = \begin{pmatrix} \langle e_1|\hat{A}|e_1\rangle & \langle e_1|\hat{A}|e_2\rangle & \cdots \\ \langle e_2|\hat{A}|e_1\rangle & \langle e_2|\hat{A}|e_2\rangle & \cdots \\ \vdots & \vdots & \ddots \end{pmatrix}$$

## 4. Operadores y Observables {#operadores}

### Clasificación de Operadores

#### Operadores Lineales

Un operador $\hat{A}$ es lineal si:
$$\hat{A}(c_1|\psi_1\rangle + c_2|\psi_2\rangle) = c_1\hat{A}|\psi_1\rangle + c_2\hat{A}|\psi_2\rangle$$

#### Operadores Hermitianos (Auto-adjuntos)

Un operador es hermitiano si:
$$\hat{A} = \hat{A}^\dagger$$

Que equivale a:
$$\langle\psi|\hat{A}|\phi\rangle = \langle\phi|\hat{A}|\psi\rangle^*$$

**Propiedades**:
- Todos sus valores propios son reales
- Sus vectores propios (de valores propios distintos) son ortogonales
- Forman una base completa del espacio
- Representan observables físicas

#### Operadores Unitarios

Un operador $\hat{U}$ es unitario si:
$$\hat{U}^\dagger\hat{U} = \hat{U}\hat{U}^\dagger = \mathbb{I}$$

**Propiedades**:
- Preservan el producto interno: $\langle\hat{U}\psi|\hat{U}\phi\rangle = \langle\psi|\phi\rangle$
- Preservan la norma: $\|\hat{U}|\psi\rangle\| = \||\psi\rangle\|$
- Representan simetrías y transformaciones

### Operadores Importantes

#### Operador de Proyección

El proyector sobre el estado $|a\rangle$ (normalizado) es:
$$\hat{P}_a = |a\rangle\langle a|$$

Propiedades:
- $\hat{P}_a^2 = \hat{P}_a$ (idempotente)
- $\hat{P}_a^\dagger = \hat{P}_a$ (hermitiano)
- Proyecta cualquier vector sobre $|a\rangle$: $\hat{P}_a|\psi\rangle = \langle a|\psi\rangle|a\rangle$

#### Operador Identidad

Si $\{|e_i\rangle\}$ es una base ortonormal:
$$\mathbb{I} = \sum_i |e_i\rangle\langle e_i|$$

#### Relación de Conmutación Canónica

Entre los operadores de posición y momento:
$$[\hat{x}, \hat{p}] = \hat{x}\hat{p} - \hat{p}\hat{x} = i\hbar$$

Más generalmente, en tres dimensiones:
$$[\hat{x}_i, \hat{p}_j] = i\hbar\delta_{ij}$$
$$[\hat{x}_i, \hat{x}_j] = 0, \quad [\hat{p}_i, \hat{p}_j] = 0$$

### Valor Esperado de un Observable

Dado un observable $\hat{A}$ y un estado $|\psi\rangle$ (normalizado), el **valor esperado** es:
$$\langle A \rangle = \langle\psi|\hat{A}|\psi\rangle$$

En la base de valores propios de $\hat{A}$:
$$\langle A \rangle = \sum_n a_n P(a_n) = \sum_n a_n |c_n|^2$$

### Desviación Estándar

La incertidumbre o dispersión es:
$$\Delta A = \sqrt{\langle A^2 \rangle - \langle A \rangle^2}$$

donde $\langle A^2 \rangle = \langle\psi|\hat{A}^2|\psi\rangle$.

### Desigualdad de Incertidumbre Generalizada

Para dos observables $\hat{A}$ y $\hat{B}$:
$$\Delta A \cdot \Delta B \geq \frac{1}{2}|\langle[\hat{A}, \hat{B}]\rangle|$$

Casos particulares:
- Posición y momento: $\Delta x \cdot \Delta p \geq \frac{\hbar}{2}$
- Energía y tiempo: $\Delta E \cdot \Delta t \geq \frac{\hbar}{2}$

## 5. Experimento de Stern-Gerlach {#stern-gerlach}

Este es un experimento fundamental que ilustra los principios de la medición cuántica y el spin.

### Descripción del Experimento

Un haz de partículas con spin se envía a través de:
1. Un imán de Stern-Gerlach orientado en la dirección $\hat{z}$
2. El haz se divide en dos componentes (spin-up $|+\rangle$ y spin-down $|-\rangle$)
3. Se selecciona una componente (ej. $|+\rangle$)
4. Se envía a través de otro imán de Stern-Gerlach en una dirección diferente

### Espacio de Hilbert para Spin-1/2

Es el espacio 2-dimensional $\mathbb{C}^2$. Usamos la notación estándar:

$$|\uparrow\rangle \equiv |+\rangle_z = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \quad |\downarrow\rangle \equiv |-\rangle_z = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$

### Estados de Spin en Diferentes Direcciones

Para spin-1/2 a lo largo de la dirección $\hat{n} = (\sin\theta\cos\phi, \sin\theta\sin\phi, \cos\theta)$:

**Componente spin-up**:
$$|+\rangle_n = \cos\frac{\theta}{2}|+\rangle_z + e^{i\phi}\sin\frac{\theta}{2}|-\rangle_z$$

**Componente spin-down**:
$$|-\rangle_n = -\sin\frac{\theta}{2}|+\rangle_z + e^{i\phi}\cos\frac{\theta}{2}|-\rangle_z$$

### Operador de Spin

El operador de spin se define como:
$$\hat{S}_i = \frac{\hbar}{2}\sigma_i \quad (i = x, y, z)$$

donde $\sigma_i$ son las **matrices de Pauli**:

$$\sigma_x = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \quad \sigma_y = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \quad \sigma_z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

**Valores propios**: $\pm\frac{\hbar}{2}$

### Probabilidades de Transición

Si preparamos el sistema en estado $|+\rangle_z$ y medimos en dirección $\hat{n}$:

$$P(+|\hat{n}) = |\langle +|\hat{n}\rangle\langle \hat{n}|+\rangle_z|^2 = \cos^2\frac{\theta}{2}$$

$$P(-|\hat{n}) = \sin^2\frac{\theta}{2}$$

### Relaciones de Conmutación del Spin

$$[\hat{S}_i, \hat{S}_j] = i\hbar\epsilon_{ijk}\hat{S}_k$$

donde $\epsilon_{ijk}$ es el símbolo de Levi-Civita.

En particular:
$$[\hat{S}_x, \hat{S}_y] = i\hbar\hat{S}_z, \quad [\hat{S}_y, \hat{S}_z] = i\hbar\hat{S}_x, \quad [\hat{S}_z, \hat{S}_x] = i\hbar\hat{S}_y$$

### Observación Clave

El experimento de Stern-Gerlach demuestra:
1. **Cuantización**: El spin solo toma valores discretos
2. **Incompatibilidad de observables**: No podemos medir dos componentes de spin simultáneamente
3. **Superposición cuántica**: Un estado que es definido en una dirección es indefinido en otra

## 6. Principio de Superposición {#superposicion}

### Enunciado del Principio

Si $|\psi_1\rangle$ y $|\psi_2\rangle$ son estados posibles de un sistema, entonces cualquier combinación lineal:
$$|\psi\rangle = c_1|\psi_1\rangle + c_2|\psi_2\rangle$$

con $c_1, c_2 \in \mathbb{C}$ también es un estado posible del sistema.

**Consecuencia**: El espacio de estados es un espacio vectorial lineal (espacio de Hilbert).

### Generalización

Para un conjunto completo de estados $\{|a_n\rangle\}$:
$$|\psi\rangle = \sum_n c_n|a_n\rangle$$

### Superposiciones Coherentes vs Mezclas Estadísticas

#### Superposición Coherente

Un estado de superposición como:
$$|\psi\rangle = \frac{1}{\sqrt{2}}(|+\rangle_z + |-\rangle_z)$$

es un estado definido único. Las probabilidades:
$$P(+) = |\frac{1}{\sqrt{2}}|^2 = \frac{1}{2}, \quad P(-) = |\frac{1}{\sqrt{2}}|^2 = \frac{1}{2}$$

resultan de **interferencias cuánticas** entre los términos.

#### Mezcla Estadística

Una mezcla estadística donde 50% de las veces el sistema está en $|+\rangle_z$ y 50% en $|-\rangle_z$ da las mismas probabilidades marginales, pero no es el mismo estado.

La diferencia es observable cuando se mide en una base diferente.

### Ejemplo: Interferencia Cuántica

Considere:
$$|\psi_1\rangle = \frac{1}{\sqrt{2}}(|+\rangle_z + |-\rangle_z)$$
$$|\psi_2\rangle = \frac{1}{\sqrt{2}}(|+\rangle_z - |-\rangle_z)$$

Ambos tienen probabilidades iguales al medir en $z$, pero en la base $x$:

Usando $|+\rangle_x = \frac{1}{\sqrt{2}}(|+\rangle_z + |-\rangle_z)$ y $|-\rangle_x = \frac{1}{\sqrt{2}}(|+\rangle_z - |-\rangle_z)$:

$$|\psi_1\rangle = |+\rangle_x \Rightarrow P_x(+) = 1, P_x(-) = 0$$
$$|\psi_2\rangle = |-\rangle_x \Rightarrow P_x(+) = 0, P_x(-) = 1$$

**Las fases relativas en la superposición son observables físicamente.**

### Fases Globales vs Locales

**Fase global**: Multiplicar todo el estado por $e^{i\alpha}$ no cambia nada:
$$|\psi\rangle \sim e^{i\alpha}|\psi\rangle \quad (\text{mismo estado físico})$$

**Fases relativas**: Son observables. Por ejemplo:
$$|\psi\rangle = c_1|\psi_1\rangle + c_2 e^{i\delta}|\psi_2\rangle$$

El factor $e^{i\delta}$ es físicamente significativo.

## 7. Medición Cuántica {#medicion}

### Teoría de la Medición

La medición es un aspecto central y controversial de la mecánica cuántica.

### Proceso de Medición Ideal (Proyectivo)

Consideremos medir un observable $\hat{A}$ con espectro discreto $\{a_n\}$:

1. **Antes de la medición**: Sistema en estado $|\psi\rangle$

2. **Durante la medición**: El aparato interactúa con el sistema

3. **Resultado**: Se obtiene valor $a_n$ con probabilidad
   $$P(a_n) = \langle\psi|\hat{P}_n|\psi\rangle$$
   
   donde $\hat{P}_n = \sum_{i: a_i = a_n} |a_i\rangle\langle a_i|$ es el proyector al subespacio de degeneración $a_n$.

4. **Después de la medición**: Sistema en estado colapsado
   $$|\psi'\rangle = \frac{\hat{P}_n|\psi\rangle}{\sqrt{\langle\psi|\hat{P}_n|\psi\rangle}}$$

### Mediciones Consecutivas

**Teorema**: Si medimos $\hat{A}$ dos veces consecutivas, la segunda medición produce el mismo resultado (ideal).

Esto es una consecuencia del colapso.

### Compatibilidad de Observables

Dos observables $\hat{A}$ y $\hat{B}$ son **compatibles** si comparten la misma base de vectores propios, lo que equivale a:
$$[\hat{A}, \hat{B}] = 0$$

**Propiedades**:
- Observables compatibles tienen valores simultáneamente definidos
- Pueden medirse simultáneamente sin perturbación mutua
- Sus dispersiones no están limitadas por relaciones de incertidumbre

### Incompatibilidad y Relaciones de Incertidumbre

Si $[\hat{A}, \hat{B}] \neq 0$, los observables son **incompatibles**:

$$\Delta A \cdot \Delta B \geq \frac{1}{2}|\langle[\hat{A}, \hat{B}]\rangle|$$

**Ejemplo**: Posición y momento con $[\hat{x}, \hat{p}] = i\hbar$:

$$\langle[\hat{x}, \hat{p}]\rangle = i\hbar$$
$$\Delta x \cdot \Delta p \geq \frac{\hbar}{2}$$

### Estados de Incertidumbre Mínima

Para $\Delta x \cdot \Delta p = \frac{\hbar}{2}$ (mínimo permitido), los estados se llaman **estados de incertidumbre mínima** o **paquetes de onda gaussianos**.

El estado fundamental del oscilador armónico es un ejemplo.

### Perturbación por Medición

Medir un observable perturba aquellos observables incompatibles con él.

**Ejemplo**: 
- Si medimos $S_z$ en $|+\rangle_z$, obtenemos $+\hbar/2$
- El colapso es a $|+\rangle_z$
- Si ahora medimos $S_x$:
  $$P(+\hbar/2) = |\langle +|_x|+\rangle_z|^2 = \frac{1}{2}$$
  $$P(-\hbar/2) = |\langle -|_x|+\rangle_z|^2 = \frac{1}{2}$$

La medición anterior se ha "olvidado" debido a la incompatibilidad.

## 8. Evolución Temporal y Ecuación de Schrödinger {#schrodinger}

### Ecuación de Schrödinger Dependiente del Tiempo

$$i\hbar\frac{\partial|\psi(t)\rangle}{\partial t} = \hat{H}|\psi(t)\rangle$$

donde $\hat{H}$ es el **operador Hamiltoniano** (energía total).

### Ecuación de Schrödinger Independiente del Tiempo

Para estados estacionarios (energía definida), buscamos soluciones de la forma:
$$|\psi(t)\rangle = |\psi\rangle e^{-iEt/\hbar}$$

La ecuación de Schrödinger independiente del tiempo es:
$$\hat{H}|\psi\rangle = E|\psi\rangle$$

donde $E$ es la energía (valor propio).

### Hamiltoniano para Partícula en Potencial

Para una partícula en potencial $V(\vec{r})$:
$$\hat{H} = \frac{\hat{p}^2}{2m} + V(\vec{r})$$

$$\hat{H} = -\frac{\hbar^2}{2m}\nabla^2 + V(\vec{r})$$

En representación de posición, la ecuación de Schrödinger es:
$$-\frac{\hbar^2}{2m}\nabla^2\psi(\vec{r}, t) + V(\vec{r})\psi(\vec{r}, t) = i\hbar\frac{\partial\psi(\vec{r}, t)}{\partial t}$$

### Operador de Evolución Temporal

La solución formal es:
$$|\psi(t)\rangle = \hat{U}(t)|\psi(0)\rangle$$

donde el **operador de evolución** es:
$$\hat{U}(t) = e^{-i\hat{H}t/\hbar}$$

Propiedades:
- Es unitario: $\hat{U}^\dagger(t)\hat{U}(t) = \mathbb{I}$
- Es composable: $\hat{U}(t_1 + t_2) = \hat{U}(t_1)\hat{U}(t_2)$
- Preserva la norma: $\langle\psi(t)|\psi(t)\rangle = \langle\psi(0)|\psi(0)\rangle$

### Conservación de la Probabilidad

La ecuación de continuidad para la densidad de probabilidad $\rho(\vec{r}, t) = |\psi(\vec{r}, t)|^2$ es:

$$\frac{\partial\rho}{\partial t} + \nabla\cdot\vec{J} = 0$$

donde la **corriente de probabilidad** es:
$$\vec{J} = \frac{\hbar}{2mi}(\psi^*\nabla\psi - \psi\nabla\psi^*)$$

o equivalentemente:
$$\vec{J} = \text{Re}\left(\psi^*\frac{\hbar}{m i}\nabla\psi\right)$$

### Conservación de Valores Esperados

Para un observable $\hat{A}$ que no depende explícitamente del tiempo:

$$\frac{d\langle A\rangle}{dt} = \frac{1}{i\hbar}\langle[\hat{A}, \hat{H}]\rangle$$

Este es el **Teorema de Ehrenfest**.

**Corolario**: Si $[\hat{A}, \hat{H}] = 0$, entonces $\langle A\rangle$ es constante (conservado).

### Ejemplos Importantes

#### Partícula Libre
$$\hat{H} = \frac{\hat{p}^2}{2m}$$

Estados propios: ondas planas $e^{i\vec{k}\cdot\vec{r}}$ con $E = \frac{\hbar^2k^2}{2m}$ (espectro continuo)

#### Oscilador Armónico
$$\hat{H} = \frac{\hat{p}^2}{2m} + \frac{1}{2}m\omega^2\hat{x}^2$$

Energías: $E_n = \hbar\omega(n + \frac{1}{2})$, $n = 0, 1, 2, ...$

## 9. Momento Angular {#momento-angular}

### Definición Clásica y Cuántica

En mecánica clásica:
$$\vec{L} = \vec{r} \times \vec{p}$$

En mecánica cuántica (operador):
$$\hat{\vec{L}} = \hat{\vec{r}} \times \hat{\vec{p}}$$

Con componentes:
$$\hat{L}_x = \hat{y}\hat{p}_z - \hat{z}\hat{p}_y$$
$$\hat{L}_y = \hat{z}\hat{p}_x - \hat{x}\hat{p}_z$$
$$\hat{L}_z = \hat{x}\hat{p}_y - \hat{y}\hat{p}_x$$

### Relaciones de Conmutación del Momento Angular

$$[\hat{L}_i, \hat{L}_j] = i\hbar\epsilon_{ijk}\hat{L}_k$$

Las relaciones cíclicas son:
$$[\hat{L}_x, \hat{L}_y] = i\hbar\hat{L}_z$$
$$[\hat{L}_y, \hat{L}_z] = i\hbar\hat{L}_x$$
$$[\hat{L}_z, \hat{L}_x] = i\hbar\hat{L}_y$$

### Operador de Magnitud Cuadrada

$$\hat{L}^2 = \hat{L}_x^2 + \hat{L}_y^2 + \hat{L}_z^2$$

**Propiedad crucial**:
$$[\hat{L}^2, \hat{L}_i] = 0 \quad (i = x, y, z)$$

Esto significa que $\hat{L}^2$ y una componente (usualmente $\hat{L}_z$) pueden tener valores definidos simultáneamente.

### Estados Propios del Momento Angular Orbital

Los estados propios comunes de $\hat{L}^2$ y $\hat{L}_z$ son los **armónicos esféricos** $Y_{\ell m}(\theta, \phi)$:

$$\hat{L}^2 Y_{\ell m} = \hbar^2\ell(\ell + 1) Y_{\ell m}$$
$$\hat{L}_z Y_{\ell m} = \hbar m Y_{\ell m}$$

donde:
- $\ell = 0, 1, 2, 3, ...$ (número cuántico orbital)
- $m = -\ell, -\ell+1, ..., \ell-1, \ell$ (número cuántico magnético)

**Valores posibles**:
- $L^2 = \hbar^2\ell(\ell + 1)$
- $L_z = \hbar m$

### Operadores de Escalera

Definimos los operadores:
$$\hat{L}_\pm = \hat{L}_x \pm i\hat{L}_y$$

**Propiedades**:
$$[\hat{L}_z, \hat{L}_\pm] = \pm\hbar\hat{L}_\pm$$
$$[\hat{L}_+, \hat{L}_-] = 2\hbar\hat{L}_z$$

**Acciones**:
$$\hat{L}_+ |\ell, m\rangle = \hbar\sqrt{\ell(\ell+1) - m(m+1)} |\ell, m+1\rangle$$
$$\hat{L}_- |\ell, m\rangle = \hbar\sqrt{\ell(\ell+1) - m(m-1)} |\ell, m-1\rangle$$

### Spin

El spin es momento angular intrínseco, sin análogo clásico. Para spin-1/2:

$$\hat{S}_i = \frac{\hbar}{2}\sigma_i$$

donde $\sigma_i$ son las matrices de Pauli.

**Valores propios de $\hat{S}^2$**:
$$\hat{S}^2|s\rangle = \hbar^2 s(s+1)|s\rangle$$

Para spin-1/2: $s = 1/2$, entonces $S^2 = \frac{3\hbar^2}{4}$

**Valores propios de $\hat{S}_z$**:
$$S_z = \pm\frac{\hbar}{2}$$

## 10. Aplicaciones y Ejemplos {#aplicaciones}

### Partícula en una Caja 1D

**Potencial**:
$$V(x) = \begin{cases} 0 & 0 < x < a \\ \infty & \text{en otro lugar} \end{cases}$$

**Ecuación de Schrödinger independiente del tiempo** (región $0 < x < a$):
$$-\frac{\hbar^2}{2m}\frac{d^2\psi}{dx^2} = E\psi$$

**Condiciones de frontera**: $\psi(0) = \psi(a) = 0$

**Soluciones**:
$$\psi_n(x) = \sqrt{\frac{2}{a}}\sin\frac{n\pi x}{a}, \quad n = 1, 2, 3, ...$$

**Energías**:
$$E_n = \frac{n^2\pi^2\hbar^2}{2ma^2} = \frac{n^2h^2}{8ma^2}$$

**Propiedades**:
- Espectro discreto (cuantización)
- Niveles no degenerados
- Separación de niveles decrece con $a^2$
- Energía del estado fundamental nunca es cero (principio de incertidumbre)

### Oscilador Armónico Cuántico

**Hamiltoniano**:
$$\hat{H} = \frac{\hat{p}^2}{2m} + \frac{1}{2}m\omega^2\hat{x}^2$$

**Operadores de Escalera**:  
Definimos:
$$\hat{a} = \sqrt{\frac{m\omega}{2\hbar}}\hat{x} + i\sqrt{\frac{1}{2m\hbar\omega}}\hat{p}$$
$$\hat{a}^\dagger = \sqrt{\frac{m\omega}{2\hbar}}\hat{x} - i\sqrt{\frac{1}{2m\hbar\omega}}\hat{p}$$

**Relaciones de conmutación**:
$$[\hat{a}, \hat{a}^\dagger] = 1$$

**Hamiltoniano en términos de escaleras**:
$$\hat{H} = \hbar\omega\left(\hat{a}^\dagger\hat{a} + \frac{1}{2}\right) = \hbar\omega\left(\hat{N} + \frac{1}{2}\right)$$

donde $\hat{N} = \hat{a}^\dagger\hat{a}$ es el operador de número.

**Estados propios**:
$$|n\rangle = \frac{(\hat{a}^\dagger)^n}{\sqrt{n!}}|0\rangle$$

**Energías**:
$$E_n = \hbar\omega\left(n + \frac{1}{2}\right), \quad n = 0, 1, 2, ...$$

**Acciones de operadores**:
$$\hat{a}|n\rangle = \sqrt{n}|n-1\rangle$$
$$\hat{a}^\dagger|n\rangle = \sqrt{n+1}|n+1\rangle$$

### Átomo de Hidrógeno

**Hamiltoniano** (sin spin):
$$\hat{H} = \frac{\hat{p}^2}{2m_e} - \frac{e^2}{4\pi\epsilon_0 r}$$

**Estados propios**: Caracterizados por números cuánticos $(n, \ell, m)$

**Energías** (Fórmula de Rydberg):
$$E_n = -\frac{13.6 \text{ eV}}{n^2}$$

donde $n = 1, 2, 3, ...$ (número cuántico principal).

**Números cuánticos**:
- $n = 1, 2, 3, ...$ (principal)
- $\ell = 0, 1, ..., n-1$ (orbital)
- $m_\ell = -\ell, ..., \ell$ (magnético)
- $m_s = ±1/2$ (spin)

**Funciones de onda del estado fundamental** ($n=1, \ell=0, m=0$):
$$\psi_{100}(r) = \frac{1}{\sqrt{\pi}a_0^{3/2}}e^{-r/a_0}$$

donde $a_0 = \frac{4\pi\epsilon_0\hbar^2}{m_e e^2} \approx 0.529$ Å es el radio de Bohr.

### Transiciones y Reglas de Selección

Las transiciones entre estados ocurren por absorción o emisión de fotones.

**Frecuencia de transición**:
$$\nu = \frac{E_f - E_i}{h}$$

**Reglas de selección** (para transiciones dipolares):
- $\Delta n$: cualquier valor
- $\Delta \ell = \pm 1$
- $\Delta m = 0, \pm 1$

Las transiciones prohibidas por estas reglas tienen probabilidad muchísimo menor.
