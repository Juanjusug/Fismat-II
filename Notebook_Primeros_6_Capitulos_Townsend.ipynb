## Capitulo 1 - Experimentos de Stern-Gerlach

### Idea central
Un haz de atomos con momento magnetico (por el espin del electron) pasa por un campo magnetico no uniforme y se separa en trayectorias discretas.

Eso muestra algo muy fuerte: el espin medido en un eje (por ejemplo z) no toma cualquier valor, solo dos para espin 1/2: +\hbar/2 o -\hbar/2.

### Conceptos clave
- Cuantizacion: resultados discretos de una medicion.
- Estado de espin: informacion cuantica del sistema.
- Medir en ejes distintos (x, y, z) cambia probabilidades.

### Ecuaciones importantes y significado
1. Energia de dipolo magnetico en campo:
$$U = -\vec{\mu}\cdot\vec{B}$$
Si el dipolo esta alineado, baja energia; si esta anti-alineado, sube.

2. Fuerza en campo no uniforme (direccion z):
$$F_z = \frac{\partial}{\partial z}(\vec{\mu}\cdot\vec{B}) \approx \mu_z\frac{\partial B_z}{\partial z}$$
Por eso el haz se desvía hacia arriba o abajo.

3. Relacion momento magnetico-espin (electron):
$$\vec{\mu} = -g\,\mu_B\,\frac{\vec{S}}{\hbar}, \qquad \mu_B = \frac{e\hbar}{2m_e}$$
Con espin 1/2 aparecen dos resultados en cada eje.

4. Valores posibles de medicion de espin en z:
$$S_z\,|+z\rangle = +\frac{\hbar}{2}|+z\rangle, \qquad S_z\,|-z\rangle = -\frac{\hbar}{2}|-z\rangle$$

### Resumen de ecuaciones del Capitulo 1
$$U = -\vec{\mu}\cdot\vec{B}$$
$$F_z \approx \mu_z\frac{\partial B_z}{\partial z}$$
$$\vec{\mu} = -g\,\mu_B\,\frac{\vec{S}}{\hbar}$$
$$\mu_B = \frac{e\hbar}{2m_e}$$
$$S_z\,|\pm z\rangle = \pm\frac{\hbar}{2}|\pm z\rangle$$

## Capitulo 2 - Rotacion de bases y mecanica matricial

### Idea central
El mismo estado cuantico puede escribirse en distintas bases (por ejemplo base z o base x). Cambiar de base no cambia la fisica; cambia la forma de escribirla.

### Conceptos clave
- Un estado es un vector |\psi\rangle en un espacio complejo.
- Un observable (como espin) se representa por un operador hermitico.
- En base concreta, estados y operadores se vuelven vectores y matrices.

### Ecuaciones importantes
1. Expansion en base ortonormal \{|n\rangle\}:
$$|\psi\rangle = \sum_n c_n |n\rangle, \qquad c_n = \langle n|\psi\rangle$$

2. Condicion de normalizacion:
$$\langle \psi|\psi\rangle = 1 \quad \Longrightarrow \quad \sum_n |c_n|^2 = 1$$

3. Probabilidad de medir el autovalor asociado a |n\rangle:
$$P_n = |\langle n|\psi\rangle|^2$$

4. Operadores de espin para espin 1/2 (matrices de Pauli):
$$S_x = \frac{\hbar}{2}\sigma_x,\; S_y = \frac{\hbar}{2}\sigma_y,\; S_z = \frac{\hbar}{2}\sigma_z$$
$$\sigma_x = \begin{pmatrix}0&1\\1&0\end{pmatrix},\;\sigma_y=\begin{pmatrix}0&-i\\i&0\end{pmatrix},\;\sigma_z=\begin{pmatrix}1&0\\0&-1\end{pmatrix}$$

5. Cambio de base por matriz unitaria U:
$$|\psi\rangle_{\text{nueva}} = U|\psi\rangle_{\text{vieja}}, \qquad U^\dagger U = I$$

### Resumen de ecuaciones del Capitulo 2
$$|\psi\rangle = \sum_n c_n |n\rangle, \; c_n = \langle n|\psi\rangle$$
$$\sum_n |c_n|^2 = 1$$
$$P_n = |\langle n|\psi\rangle|^2$$
$$S_i = \frac{\hbar}{2}\sigma_i \;(i=x,y,z)$$
$$U^\dagger U = I$$

## Capitulo 3 - Momento angular

### Idea central
El momento angular cuantico (orbital y espin) esta cuantizado. No puedes fijar simultaneamente todas sus componentes con precision absoluta.

### Conceptos clave
- Operadores: L_x,L_y,L_z o en general J_x,J_y,J_z.
- Magnitud total J^2 = J_x^2+J_y^2+J_z^2.
- Numeros cuanticos j,m.

### Ecuaciones importantes
1. Conmutadores de momento angular:
$$[J_x,J_y]=i\hbar J_z,\; [J_y,J_z]=i\hbar J_x,\; [J_z,J_x]=i\hbar J_y$$

2. Autovalores de J^2 y J_z:
$$J^2|j,m\rangle = \hbar^2 j(j+1)|j,m\rangle$$
$$J_z|j,m\rangle = \hbar m|j,m\rangle$$
con m=-j,-j+1,\dots,j.

3. Operadores de escalera:
$$J_\pm = J_x \pm iJ_y$$
$$J_\pm|j,m\rangle = \hbar\sqrt{j(j+1)-m(m\pm1)}\,|j,m\pm1\rangle$$

4. Incertidumbre entre componentes:
$$\Delta J_x\,\Delta J_y \ge \frac{1}{2}|\langle[J_x,J_y]\rangle| = \frac{\hbar}{2}|\langle J_z\rangle|$$

### Resumen de ecuaciones del Capitulo 3
$$[J_x,J_y]=i\hbar J_z,\;[J_y,J_z]=i\hbar J_x,\;[J_z,J_x]=i\hbar J_y$$
$$J^2|j,m\rangle = \hbar^2 j(j+1)|j,m\rangle$$
$$J_z|j,m\rangle = \hbar m|j,m\rangle$$
$$J_\pm = J_x \pm iJ_y$$
$$J_\pm|j,m\rangle = \hbar\sqrt{j(j+1)-m(m\pm1)}\,|j,m\pm1\rangle$$
$$\Delta J_x\,\Delta J_y \ge \frac{\hbar}{2}|\langle J_z\rangle|$$

## Capitulo 4 - Evolucion temporal

### Idea central
La funcion de onda/estado cambia en el tiempo segun la ecuacion de Schrodinger. El Hamiltoniano H gobierna completamente esa evolucion.

### Conceptos clave
- Si H no depende del tiempo, los autoestados de energia solo ganan una fase temporal.
- Superposiciones producen oscilaciones en probabilidades.

### Ecuaciones importantes
1. Ecuacion de Schrodinger dependiente del tiempo:
$$i\hbar\frac{d}{dt}|\psi(t)\rangle = H|\psi(t)\rangle$$

2. Operador de evolucion (si H constante):
$$|\psi(t)\rangle = e^{-iHt/\hbar}|\psi(0)\rangle$$

3. Para autoestado |n\rangle con energia E_n:
$$|n,t\rangle = e^{-iE_nt/\hbar}|n,0\rangle$$

4. Valor esperado de observable A:
$$\langle A \rangle_t = \langle \psi(t)|A|\psi(t)\rangle$$

5. Ecuacion de Ehrenfest (forma general):
$$\frac{d}{dt}\langle A\rangle = \frac{i}{\hbar}\langle[H,A]\rangle + \left\langle\frac{\partial A}{\partial t}\right\rangle$$

### Resumen de ecuaciones del Capitulo 4
$$i\hbar\frac{d}{dt}|\psi(t)\rangle = H|\psi(t)\rangle$$
$$|\psi(t)\rangle = e^{-iHt/\hbar}|\psi(0)\rangle$$
$$|n,t\rangle = e^{-iE_nt/\hbar}|n,0\rangle$$
$$\langle A \rangle_t = \langle \psi(t)|A|\psi(t)\rangle$$
$$\frac{d}{dt}\langle A\rangle = \frac{i}{\hbar}\langle[H,A]\rangle + \left\langle\frac{\partial A}{\partial t}\right\rangle$$

## Capitulo 5 - Sistema de dos particulas de espin 1/2

### Idea central
Cuando tienes dos espines 1/2, el espacio total es el producto tensorial de ambos. Aparecen estados acoplados: triplete (simetrico) y singlete (antisimetrico).

### Conceptos clave
- Base producto: |+\rangle|+\rangle, |+\rangle|-\rangle, |-\rangle|+\rangle, |-\rangle|-\rangle.
- Base acoplada: |1,1\rangle, |1,0\rangle, |1,-1\rangle, |0,0\rangle.
- Entrelazamiento: no se puede separar en estado de part 1 por estado de part 2.

### Ecuaciones importantes
1. Espacio total:
$$\mathcal{H}_{\text{total}} = \mathcal{H}_1 \otimes \mathcal{H}_2$$

2. Operador espin total:
$$\vec{S} = \vec{S}_1 + \vec{S}_2$$
$$S^2 = S_1^2 + S_2^2 + 2\,\vec{S}_1\cdot\vec{S}_2$$

3. Estados triplete y singlete:
$$|1,1\rangle = |+\rangle|+\rangle$$
$$|1,0\rangle = \frac{1}{\sqrt{2}}\left(|+\rangle|-\rangle + |-\rangle|+\rangle\right)$$
$$|1,-1\rangle = |-\rangle|-\rangle$$
$$|0,0\rangle = \frac{1}{\sqrt{2}}\left(|+\rangle|-\rangle - |-\rangle|+\rangle\right)$$

4. Autovalores de S^2:
$$S^2|s,m\rangle = \hbar^2 s(s+1)|s,m\rangle$$
con s=1 (triplete) o s=0 (singlete).

### Resumen de ecuaciones del Capitulo 5
$$\mathcal{H}_{\text{total}} = \mathcal{H}_1 \otimes \mathcal{H}_2$$
$$\vec{S} = \vec{S}_1 + \vec{S}_2$$
$$S^2 = S_1^2 + S_2^2 + 2\,\vec{S}_1\cdot\vec{S}_2$$
$$|1,1\rangle = |+\rangle|+\rangle$$
$$|1,0\rangle = \frac{1}{\sqrt{2}}\left(|+\rangle|-\rangle + |-\rangle|+\rangle\right)$$
$$|1,-1\rangle = |-\rangle|-\rangle$$
$$|0,0\rangle = \frac{1}{\sqrt{2}}\left(|+\rangle|-\rangle - |-\rangle|+\rangle\right)$$
$$S^2|s,m\rangle = \hbar^2 s(s+1)|s,m\rangle$$

## Capitulo 6 - Ejercicios resueltos y explicados (Mecanica ondulatoria 1D)

Convenciones usadas:

- Operador momento: $\hat{p}=-i\hbar\,\partial_x$
- Ecuacion de Schrodinger 1D:
$$i\hbar\frac{\partial\psi}{\partial t}=\left[-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial x^2}+V(x)\right]\psi$$
- Densidad de probabilidad: $\rho=|\psi|^2$
- Corriente de probabilidad:
$$j=\frac{\hbar}{2mi}\left(\psi^*\frac{\partial\psi}{\partial x}-\psi\frac{\partial\psi^*}{\partial x}\right)$$

Estructura:

2. Resolucion paso a paso
3. Resultado final

### Ejercicio 6.1 - Autofunciones de momento y normalizacion delta

**Problema reformulado**
Verifica que una onda plana $\phi_k(x)=A e^{ikx}$ es autofuncion de $\hat{p}$ y determina su normalizacion en el continuo.

**Paso 1: aplicar el operador momento**
$$\hat{p}\phi_k=-i\hbar\frac{d}{dx}(Ae^{ikx})=-i\hbar(ik)Ae^{ikx}=\hbar k\,\phi_k$$
Entonces el autovalor es:
$$p=\hbar k$$

**Paso 2: por que no se normaliza a 1 en todo el eje**
Como $|e^{ikx}|^2=1$,
$$\int_{-\infty}^{\infty}|\phi_k|^2dx=|A|^2\int_{-\infty}^{\infty}dx=\infty$$
No es un estado cuadrado-integrable; se usa normalizacion delta.

**Paso 3: imponer ortonormalidad en continuo**
Queremos
$$\langle k|k'\rangle=\int_{-\infty}^{\infty}\phi_k^*(x)\phi_{k'}(x)dx=\delta(k-k')$$
Con
$$\phi_k(x)=\frac{1}{\sqrt{2\pi}}e^{ikx}$$
se cumple porque
$$\int e^{-ikx}e^{ik'x}dx=2\pi\,\delta(k-k')$$

**Resultado final**
- Las ondas planas son autoestados de momento con $p=\hbar k$.
- Base continua estandar:
$$\phi_k(x)=\frac{1}{\sqrt{2\pi}}e^{ikx}$$

**Intuicion fisica**
Un estado de momento exacto esta totalmente deslocalizado en posicion: momento perfectamente definido implica incertidumbre espacial maxima.

### Ejercicio 6.2 - Ecuacion de continuidad

**Problema reformulado**
Partiendo de Schrodinger y su conjugada compleja, deduce
$$\frac{\partial\rho}{\partial t}+\frac{\partial j}{\partial x}=0$$
e identifica $\rho$ y $j$.

**Paso 1: ecuacion y conjugada**
$$i\hbar\,\partial_t\psi=-\frac{\hbar^2}{2m}\partial_x^2\psi+V\psi$$
$$-i\hbar\,\partial_t\psi^*=-\frac{\hbar^2}{2m}\partial_x^2\psi^*+V\psi^*$$

**Paso 2: combinacion util**
Multiplica la primera por $\psi^*$ y la segunda por $\psi$, luego resta:
$$i\hbar(\psi^*\partial_t\psi+\psi\partial_t\psi^*)=-\frac{\hbar^2}{2m}(\psi^*\partial_x^2\psi-\psi\partial_x^2\psi^*)$$

Lado izquierdo:
$$\psi^*\partial_t\psi+\psi\partial_t\psi^*=\partial_t(|\psi|^2)=\partial_t\rho$$

Lado derecho como derivada total:
$$\psi^*\partial_x^2\psi-\psi\partial_x^2\psi^*=\partial_x\left(\psi^*\partial_x\psi-\psi\partial_x\psi^*\right)$$

Entonces:
$$\partial_t\rho+\partial_x\left[\frac{\hbar}{2mi}\left(\psi^*\partial_x\psi-\psi\partial_x\psi^*\right)\right]=0$$

**Resultado final**
$$\rho=|\psi|^2, \qquad j=\frac{\hbar}{2mi}\left(\psi^*\partial_x\psi-\psi\partial_x\psi^*\right)$$

**Intuicion fisica**
La probabilidad no desaparece ni aparece: solo fluye de una region a otra, como un fluido conservativo.

### Ejercicio 6.3 - Paquete gaussiano e incertidumbre minima

**Problema reformulado**
Para
$$\psi(x)=\frac{1}{(2\pi\sigma_x^2)^{1/4}}\exp\left[-\frac{(x-x_0)^2}{4\sigma_x^2}\right]e^{ik_0x}$$
calcula $\Delta x$, $\Delta p$ y verifica el limite minimo.

**Paso 1: dispersion en posicion**
Para la gaussiana estandar dada,
$$\Delta x=\sigma_x$$

**Paso 2: pasar a espacio de momentos**
La transformada de Fourier de una gaussiana es gaussiana. Su ancho en $k$ es:
$$\Delta k=\frac{1}{2\sigma_x}$$
Luego
$$\Delta p=\hbar\Delta k=\frac{\hbar}{2\sigma_x}$$

**Paso 3: producto de incertidumbres**
$$\Delta x\,\Delta p=\sigma_x\frac{\hbar}{2\sigma_x}=\frac{\hbar}{2}$$

**Resultado final**
El paquete gaussiano es un estado de incertidumbre minima.

**Intuicion fisica**
La gaussiana es el mejor compromiso posible entre localizacion espacial y definicion de momento.

### Ejercicio 6.4 - Particula libre: relacion de dispersion y velocidades

**Problema reformulado**
Para una onda plana de particula libre, encuentra $\omega(k)$, velocidad de fase y velocidad de grupo.

Con $V(x)=0$, Schrodinger admite
$$\psi(x,t)=Ae^{i(kx-\omega t)}$$
Sustituyendo:
$$\hbar\omega=\frac{\hbar^2k^2}{2m}$$
asi que
$$\omega(k)=\frac{\hbar k^2}{2m}$$

**Velocidad de fase**
$$v_f=\frac{\omega}{k}=\frac{\hbar k}{2m}=\frac{v}{2}$$
(donde $v=p/m=\hbar k/m$)

**Velocidad de grupo**
$$v_g=\frac{d\omega}{dk}=\frac{\hbar k}{m}=v$$

**Resultado final**
- $v_f=v/2$
- $v_g=v$

**Intuicion fisica**
La informacion y el maximo del paquete viajan con $v_g$, que coincide con la velocidad clasica de la particula.

### Ejercicio 6.5 - Escalon de potencial: coeficientes R y T

**Problema reformulado**
Un haz incidente desde la izquierda encuentra
$$V(x)=\begin{cases}0,&x<0\\V_0,&x>0\end{cases}$$
Para $E>V_0$, calcula reflexion y transmision.

**Paso 1: soluciones por region**
Region I ($x<0$):
$$\psi_I=Ae^{ik_1x}+Be^{-ik_1x},\quad k_1=\frac{\sqrt{2mE}}{\hbar}$$
Region II ($x>0$):
$$\psi_{II}=Ce^{ik_2x},\quad k_2=\frac{\sqrt{2m(E-V_0)}}{\hbar}$$

**Paso 2: continuidad en x=0**
$$A+B=C$$
$$ik_1(A-B)=ik_2C$$

De ahi:
$$\frac{B}{A}=\frac{k_1-k_2}{k_1+k_2}, \qquad \frac{C}{A}=\frac{2k_1}{k_1+k_2}$$

**Paso 3: corrientes**
$$R=\left|\frac{B}{A}\right|^2=\left(\frac{k_1-k_2}{k_1+k_2}\right)^2$$
$$T=\frac{k_2}{k_1}\left|\frac{C}{A}\right|^2=\frac{4k_1k_2}{(k_1+k_2)^2}$$

**Resultado final**
$$R+T=1$$

**Intuicion fisica**
Aunque $E>V_0$, hay reflexion cuantica por desajuste de numero de onda entre regiones.

### Ejercicio 6.6 - Tunelamiento en barrera rectangular

**Problema reformulado**
Para una barrera
$$V(x)=\begin{cases}0,&x<0\\V_0,&0\le x\le a\\0,&x>a\end{cases}$$
con $E<V_0$, estima la transmitancia.

**Paso 1: numeros de onda**
Fuera de la barrera:
$$k=\frac{\sqrt{2mE}}{\hbar}$$
Dentro de la barrera (decaimiento):
$$\kappa=\frac{\sqrt{2m(V_0-E)}}{\hbar}$$

**Paso 2: comportamiento dentro de barrera**
La solucion es evanescente:
$$\psi_{II}(x)=Fe^{\kappa x}+Ge^{-\kappa x}$$
No oscila: decae/crece exponencialmente.

**Paso 3: resultado conocido exacto**
$$T=\left[1+\frac{V_0^2\sinh^2(\kappa a)}{4E(V_0-E)}\right]^{-1}$$

**Paso 4: aproximacion barrera opaca ($\kappa a\gg1$)**
$$T\approx 16\frac{E(V_0-E)}{V_0^2}e^{-2\kappa a}$$

**Intuicion fisica**
La probabilidad de atravesar cae exponencialmente con ancho $a$ y con altura efectiva $V_0-E$.

### Ejercicio 6.7 - Pozo infinito: autoestados, energias y valores esperados

**Problema reformulado**
Para
$$V(x)=0\ (0<x<L),\qquad V=\infty\ \text{fuera}$$
encuentra $\phi_n(x)$, $E_n$, $\langle x\rangle$ y $\langle p\rangle$ en un autoestado.

**Paso 1: condicion de frontera**
$\phi(0)=\phi(L)=0$. En el interior:
$$-\frac{\hbar^2}{2m}\phi''=E\phi\Rightarrow \phi=A\sin(kx)+B\cos(kx)$$
$\phi(0)=0\Rightarrow B=0$.
$\phi(L)=0\Rightarrow \sin(kL)=0\Rightarrow k_n=n\pi/L$.

**Paso 2: normalizacion**
$$\phi_n(x)=\sqrt{\frac{2}{L}}\sin\left(\frac{n\pi x}{L}\right),\quad n=1,2,3,...$$

**Paso 3: energias**
$$E_n=\frac{\hbar^2k_n^2}{2m}=\frac{n^2\pi^2\hbar^2}{2mL^2}$$

**Paso 4: valores esperados**
Por simetria respecto del centro del pozo:
$$\langle x\rangle=\frac{L}{2}$$
y para autoestado real estacionario, corriente neta nula:
$$\langle p\rangle=0$$

**Intuicion fisica**
Cuanto mayor $n$, mas nodos y mayor curvatura de la funcion de onda, por lo tanto mayor energia cinetica.

### Ejercicio 6.8 - Superposicion en pozo infinito y oscilacion temporal

**Problema reformulado**
Si
$$\Psi(x,0)=\frac{1}{\sqrt{2}}\left(\phi_1(x)+\phi_2(x)\right)$$
halla $\Psi(x,t)$ y describe la dependencia temporal de $|\Psi|^2$.

**Paso 1: evolucion de cada autoestado**
$$\Psi(x,t)=\frac{1}{\sqrt{2}}\left(\phi_1e^{-iE_1t/\hbar}+\phi_2e^{-iE_2t/\hbar}\right)$$

**Paso 2: densidad**
$$|\Psi|^2=\frac{1}{2}\left(|\phi_1|^2+|\phi_2|^2\right)+\phi_1\phi_2\cos\left(\frac{E_2-E_1}{\hbar}t\right)$$

**Resultado final**
La densidad no es estacionaria por el termino de interferencia.
La frecuencia angular de oscilacion es
$$\omega_{12}=\frac{E_2-E_1}{\hbar}$$

**Intuicion fisica**
Estados de energia pura solo cambian fase global; una superposicion produce batidos observables en distribuciones de probabilidad.

### Ejercicio 6.9 - Pozo finito simetrico: ecuaciones trascendentes

**Problema reformulado**
Para un pozo de ancho $2a$ y profundidad $V_0$ (centrado en 0), escribe las condiciones de cuantizacion para estados ligados ($-V_0<E<0$ usando referencia exterior).

Definimos
$$k=\frac{\sqrt{2m(E+V_0)}}{\hbar},\qquad \kappa=\frac{\sqrt{-2mE}}{\hbar}$$

Por paridad, hay dos familias:

**Estados pares**
En el interior $\sim \cos(kx)$ y fuera decae exponencialmente. Continuidad en $x=a$ da:
$$k\tan(ka)=\kappa$$

**Estados impares**
En el interior $\sim \sin(kx)$ y fuera decae exponencialmente. Continuidad en $x=a$ da:
$$-k\cot(ka)=\kappa$$

**Resultado final**
Las energias permitidas se obtienen resolviendo numericamente esas ecuaciones.

**Intuicion fisica**
No cualquier energia forma estado ligado: solo las que permiten empalmar oscilacion interna con decaimiento externo.

### Ejercicio 6.10 - Potencial delta atractivo

**Problema reformulado**
Para $V(x)=-\alpha\delta(x)$ con $\alpha>0$, encuentra el estado ligado y su energia.

**Paso 1: ecuacion para x\neq0**
Como $V=0$ fuera del origen y $E<0$:
$$\phi''=\kappa^2\phi,\qquad \kappa=\frac{\sqrt{-2mE}}{\hbar}$$
La solucion normalizable es
$$\phi(x)=Ae^{-\kappa|x|}$$

**Paso 2: condicion de salto por delta**
La funcion es continua en $0$, pero la derivada salta:
$$\phi'(0^+)-\phi'(0^-)=-\frac{2m\alpha}{\hbar^2}\phi(0)$$
Con $\phi=Ae^{-\kappa|x|}$:
$$(-\kappa A)-(+\kappa A)=-\frac{2m\alpha}{\hbar^2}A$$
$$\kappa=\frac{m\alpha}{\hbar^2}$$

**Paso 3: energia**
$$E=-\frac{\hbar^2\kappa^2}{2m}=-\frac{m\alpha^2}{2\hbar^2}$$

**Resultado final**
Siempre existe exactamente un estado ligado para delta atractivo en 1D.

**Intuicion fisica**
Incluso un potencial extremadamente localizado puede atrapar una particula cuantica.

### Ejercicio 6.11 - Teorema de Ehrenfest en 1D

**Problema reformulado**
Demuestra que las medias cuanticas obedecen ecuaciones analogas a Newton:
$$\frac{d\langle x\rangle}{dt}=\frac{\langle p\rangle}{m},\qquad \frac{d\langle p\rangle}{dt}=-\left\langle\frac{dV}{dx}\right\rangle$$

**Paso 1: para posicion**
Usando
$$\frac{d\langle A\rangle}{dt}=\frac{i}{\hbar}\langle[H,A]\rangle$$
con $A=x$ y
$$H=\frac{p^2}{2m}+V(x)$$
se tiene
$$[H,x]=\frac{1}{2m}[p^2,x]=\frac{1}{2m}(p[p,x]+[p,x]p)=-\frac{i\hbar}{m}p$$
luego
$$\frac{d\langle x\rangle}{dt}=\frac{\langle p\rangle}{m}$$

**Paso 2: para momento**
Con $A=p$:
$$[H,p]=[V(x),p]=i\hbar\frac{dV}{dx}$$
por lo que
$$\frac{d\langle p\rangle}{dt}=\frac{i}{\hbar}\langle[H,p]\rangle=-\left\langle\frac{dV}{dx}\right\rangle$$

**Resultado final**
La dinamica del centro del paquete reproduce la estructura clasica.

**Intuicion fisica**
La mecanica clasica emerge como ecuacion para valores esperados, no necesariamente para cada medicion individual.
